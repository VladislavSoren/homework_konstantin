```python
# атрибуты класса
class User:
    is_human = True  # атрибут класса is_human
    name = "Standart Name"  # атрибут класса name
    age_parog = 18

    def __init__(self, age):
        self.age = age
        # self.age_parog = cls.age_parog

    @classmethod
    def print_class_info(cls):  #  cls - ссылка на класс User
        print(f"is_human - {cls.is_human}, name - {cls.name}")

    @classmethod
    def increase_age_parog(cls, value):
        cls.age_parog += value
        print("Глобальный порог изменён")

    @staticmethod
    def add_hello(name):
        print(f"Hello print {name}")
        return f"Hello return {name}"

    def increase_age_parog_for_object(self, value):
        self.age_parog += value
        response = self.add_hello("Vova")
        return response
    
    def is_adult(self):
        # if self.age >= self.age_parog:
        #     return True
        # else:
        #     return False
        return self.age >= self.age_parog
    
    
```

```python
user = User(18)
user.increase_age_parog_for_object(2)
```

```python
# атрибуты объекта класса
class User:
    is_human = True  # атрибут класса is_human
    name = "Standart Name"  # атрибут класса name

    # def __init__(self, age): pass

    def __init__(self, age):
        self.age = age  # атрибуты объекта класса
#        super().__init__()

    # метод класса
    @classmethod
    def print_class_info(cls): #  cls - ссылка на класс User
        print(f"is_human - {cls.is_human}, name - {cls.name}, age - {cls.age}")

    # метод объекта класса
    def print_object_info(self): #  self - ссылка на объект класса User, т.е. на user (user = User())
        print(f"is_human - {self.is_human}, name - {self.name}, age - {self.age}")
```

```python
# атрибуты объекта класса
class User:
    is_human = True  # атрибут класса is_human
    name = "Standart Name"  # атрибут класса name

    # конструктор класса, через него мы можем наполнять объект новыми свойствами
    def __init__(self, age):
        self.age = age  # атрибуты объекта класса

    # метод класса
    @classmethod
    def print_class_info(cls): #  cls - ссылка на класс User
        print(f"is_human - {cls.is_human}, name - {cls.name}, age - {cls.age}")

    # метод объекта класса
    def print_object_info(self): #  self - ссылка на объект класса User, т.е. на user (user = User())
        print(f"is_human - {self.is_human}, name - {self.name}, age - {self.age}")
```

```python
# мы не можем из класса обратиться к атрибуту объекта класса  
print(User.is_human)
print(User.name)
print(User.age)
```

```python
class Person:
    def __new__(cls, *args, **kwargs):
        print("1. Вызван __new__: Создание объекта")
        return super().__new__(cls)

    def __init__(self, name):
        print("2. Вызван __init__: Инициализация объекта")
        self.name = name

# Создание объекта
p = Person("Ivan")
p
```
