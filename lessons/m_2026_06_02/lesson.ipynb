```python
# Абстрактный класс
from abc import ABC, abstractmethod

class AbstractUser(ABC):
    
    @abstractmethod
    def get_name(self):
        pass
```

```python
user = AbstractUser()
```

```python
class User(AbstractUser):
    pass

user = User()
```

```python
# Интерфейс
class User(AbstractUser):
    def get_name(self):
            print("name")

user = User()
```

```python
class Admin(AbstractUser):
    def get_name(self):
            print("name")

admin = Admin()
```

```python
# стандарт (интерфейс реализованный через абстрактный класс)
class AbstractCar(ABC):
    
    @abstractmethod
    def drive(self):
        pass

    @abstractmethod
    def stop(self):
        pass
```

```python
# Конкретный чертёж по стандарту (мы должны объявить все методы из абстрактного класса)
class CarBMV(AbstractCar):
    
    def drive(self):
        pass

    def stop(self):
        pass

class CarNISSAN(AbstractCar):
    
    def drive(self):
        pass

    def stop(self):
        pass
```

```python
# кто то пытается обойти стандарт
class CarLADA(AbstractCar):
    
    def fly(self):
        pass

    def stop(self):
        pass
```

```python
# Пример паттерна "Фабричный метод" (Factory Method)

from abc import ABC, abstractmethod

# Стандарт
class Transport(ABC):
    @abstractmethod
    def deliver(self):
        pass

class Truck(Transport):
    def deliver(self):
        return "Доставка по земле (грузовик)"

class Ship(Transport):
    def deliver(self):
        return "Доставка по морю (корабль)"
    
class Plain(Transport):
    def deliver(self):
        return "Доставка по воздуху (самолёт)"

# Создатель (фабричный метод)
class Logistics(ABC):
    @abstractmethod
    def create_transport(self) -> Transport:
        pass
    
    def plan_delivery(self):
        transport = self.create_transport()
        return transport.deliver()

class RoadLogistics(Logistics):
    def create_transport(self) -> Transport:
        truck = Truck()
        return truck

class SeaLogistics(Logistics):
    def create_transport(self) -> Transport:
        return Ship()
    
class SkyLogistics(Logistics):
    def create_transport(self) -> Transport:
        return Plain()

# Использование
logistics = RoadLogistics()
print(logistics.plan_delivery())  # Доставка по земле (грузовик)

logistics = SeaLogistics()
print(logistics.plan_delivery())  # Доставка по морю (корабль)

logistics = SkyLogistics()
print(logistics.plan_delivery())  # Доставка по морю (корабль)
```

```python
# Протоколы
from typing import Protocol

# Определяем протокол: умеет крякать
class Quackable(Protocol):
    def quack(self) -> str:
        ...

# Обычный класс утки (не наследует Quackable)
class Duck:
    def quack(self) -> str:
        return "Кря!"

# А вот человек, который научился крякать
class Person:
    def quack(self) -> str:
        return "Крякаю как утка"

# Функция ожидает любой объект, следующий протоколу Quackable
def make_it_quack(thing: Quackable) -> None:
    print(thing.quack())

# Утка подходит
duck = Duck()
make_it_quack(duck)   # Кря!

# Человек тоже подходит — утиная типизация
person = Person()
make_it_quack(person) # Крякаю как утка
```
