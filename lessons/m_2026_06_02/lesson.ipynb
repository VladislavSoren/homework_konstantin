```python
from time import sleep

def fun():
    print("processing")
    sleep(1)
    print("finish")
    
```

```python
fun()
```

```python
from time import time

def timer_decorator(any_fun):
    def wrapper():
        start_time = time()
        any_fun()
        exec_time = time() - start_time
        print("Execution time: ", exec_time)
    return wrapper
```

```python
modified_fun = timer_decorator(fun)
modified_fun()
```

```python
def fun():
    print("processing")
    sleep(1)
    print("finish")

fun = timer_decorator(fun)
```

```python
@timer_decorator
def fun():
    print("processing")
    sleep(1)
    print("finish")
```

```python
fun()
```

```python
# Смысл декоратора - возможность обокатить поведению ЛЮБОЙ функции своей дополнительной логикой
# В нашем случае мы можем замерить время выполнения ЛЮБОЙ функции
```

```python
@timer_decorator
def fun1():
    print("processing")
    sleep(1)
    print("finish")

def fun2():
    print("processing")
    sleep(2)
    print("finish")

@timer_decorator
def fun3():
    print("processing")
    sleep(3)
    print("finish")
```

```python
fun1()
fun2()
fun3()
```
