```python
from time import time, sleep

def timer_decorator(any_fun):
    def wrapper():
        start_time = time()
        any_fun()  # передана из внешней области И сохранена в wrapper (это называется замыканием)
        exec_time = time() - start_time
        print("Execution time: ", exec_time)
    return wrapper
```

```python
def timer_decorator(timeout, end_message="Конец"):  # timeout - enclosing переменная для wrapper и др. вложенных  функций

    def wrapper(any_fun): # any_fun - enclosing переменная для inner_wrapper и др. вложенных  функций

        def inner_wrapper():
            start_time = time()
            any_fun()  # передана из внешней области И сохранена в wrapper (это называется замыканием)
            exec_time = time() - start_time
            print("Execution time: ", exec_time)

            if exec_time > timeout:
                print("Выполняется слишком долго")

            print(end_message)

        print("Возвращаем inner_wrapper") 

        return inner_wrapper
    
    print("Мы запоминаем timeout")  # Это и есть замыкание, т.к. функция wrapper его найдёт в этой области

    return wrapper
            
```

```python
def fun1():
    print("processing")
    sleep(1)
    print("finish")

wrapper = timer_decorator(timeout=0.5)
inner_wrapper = wrapper(fun1)
inner_wrapper()
```

```python
@timer_decorator(timeout=0.5)
def fun1():
    print("processing")
    sleep(1)
    print("finish")
```

```python
fun1()
```

```python
@timer_decorator(timeout=2, end_message="Финиш")
def fun2():
    print("processing")
    sleep(1)
    print("finish")

fun2()
```

```python
cashe = {}

def sum_numbers(number1, number2):

    key = (number1, number2)
    if key in cashe:
        return cashe[key]

    result = number1 + number2
    sleep(1)
    cashe[key] = result

    return cashe[key]
```

```python
def cashe_decorator(fun):
    
    def wrapper(*args, **kwargs):

        key = args
        if key in cashe:
            return cashe[key]

        result = fun(*args, **kwargs)
        cashe[key] = result

        return cashe[key]

    return wrapper
```

```python
cashe = {}

@cashe_decorator
def sum_numbers(number1, number2):
    sleep(1)
    return number1 + number2
```

```python
sum_numbers(1,2)
sum_numbers(1,2)
sum_numbers(1,2)
sum_numbers(1,2)
sum_numbers(1,2)
```
