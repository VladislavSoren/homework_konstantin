```python
# Наследование
class User:
    is_human = True

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    @property
    def get_name(self):
        return self.name

    def get_info(self):
        return f"{self.name} {self.age}"
    
class Admin(User):
    def __init__(self, name, age):
#        self.level = level
        super().__init__(name, age)

    def get_info(self):
        result_from_parent = super().get_info()
        return f"{result_from_parent} {self.level}"
```

```python
admin = Admin("Bob", 25, 7)
```

```python
admin.get_info()
```
